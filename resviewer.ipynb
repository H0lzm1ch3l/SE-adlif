```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load csv file from the results/hydra folder
df = pd.read_csv("/home/michael/projects/SE-adlif/results/hydra/2025-12-05/22-24-08/logs/mlp_snn/version_0/metrics.csv")
df
```

```python
col_names = ["l1_tau_u_mean", "l2_tau_d_mean", "l2_tau_t_mean", "l2_tau_u_mean", "val_acc"]
df_plot = df[col_names]
df_plot
```

```python
# fill nan values up with all prior values and put zero if no prior value exists
df_plot = df_plot.fillna(method='ffill').fillna(0)
df_plot
```

```python
# plot all the taus and the vall acc on the same lineplot  but add 
plt.figure(figsize=(10, 6))
for col in col_names:
    plt.plot(df_plot[col], label=col)
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.title("Taus and Validation Accuracy over Epochs")
plt.legend()
plt.grid()
plt.show()
```
