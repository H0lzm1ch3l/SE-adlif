```python
class Pyramidal:
    def __init__(self, alpha, beta, a, b, theta_s = 1.0, M=1, u0=0, w0=0):
        self.a = a
        self.b = b
        self.u = u0
        self.w = w0
        self.theta_s = theta_s

    def step(self, u_tm1, w_tm1, I):
        u = self.alpha * u_tm1 + (1-self.alpha) * (I - w_tm1)

        z = u > self.theta_s
        u = z * self.u0 + (1 - z) * u

        w = self.beta * w_tm1 + (1-self.alpha) * (a * u_tm1 + b * z)
        
```

```python
class fixed_frequency_spikegen:
    def __init__(self, F):
        # internal frequency, use sine based freq gen
        self.F = F
    
    def step(self, t):
        return t//self.F 
```

```python
# we generate 1 second of data with a frequency of 60Hz this means we get 60 spikes during this 1 second with a delta t of 0.001
# because we need to follow shannons theorem temporal resolution is 1kHz (100 would not be 2x what we want to sample)
spikegen = fixed_frequency_spikegen(60)
T = 1
delta_t = 0.001
steps = T * delta_t
```

```python
spike_train = []
for k in range(steps):
    spike_train.append(spikegen.step(k))
```
