```python
import torch
import matplotlib.pyplot as plt
```

```python
# Create 100 timesteps of logarithmic decay
timesteps = torch.arange(1, 101)

def logarithmic_decay(t):
    return torch.log(1 + t)
decay = logarithmic_decay(timesteps)

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(timesteps, decay, linewidth=2)
plt.xlabel('Timesteps')
plt.ylabel('Logarithmic Decay')
plt.title('Logarithmic Decay Over 100 Timesteps')
plt.grid(True, alpha=0.3)
plt.show()
```
