```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid_decay(x, slope_k, midpoint_x0):
    return 1 / (1 + np.exp(slope_k * -(x - midpoint_x0)))

def stretched_exp(x, slope_k, scale_lambda):
    return -np.exp(- (x / scale_lambda)**slope_k) + 1

# Generate X values
x = np.linspace(0, 20, 500)

# Configure parameters
# Sigmoid: Steep drop at x=10
y_sig = sigmoid_decay(x, slope_k=2.0, midpoint_x0=10)

# Stretched Exp: Flat until ~8, then drops
y_exp = stretched_exp(x, slope_k=5.0, scale_lambda=10)

plt.figure(figsize=(10, 6))
plt.plot(x, y_sig, label='Sigmoid (Midpoint=10, Slope=2)', linewidth=2)
plt.plot(x, y_exp, label='Stretched Exp (Lambda=10, k=5)', linewidth=2, linestyle='--')
plt.grid(True, alpha=0.3)
plt.legend()
plt.title("Comparison of Tunable Decay Functions")
plt.show()
```

```python
timesteps = torch.arange(-1.0, 1.0, 0.01)

def reverse_sigmoid(x, k=1.0, x0=1.0):
    return 1 / (1 + torch.exp(k * -(x - x0)))

decay_values = reverse_sigmoid(timesteps, k=25.0, x0=0.2)

# Plot the data
plt.plot(timesteps.numpy(), decay_values.numpy())
plt.title("Reverse Sigmoid Decay Function")
plt.xlabel("Time")
plt.ylabel("Decay Value")
plt.grid()
plt.show()
```

```python
# Flexible decay function with adjustable time constant and exponent
def decay_function(t, tau=1.0, exponent=1.0, curve_type='power'):
    """
    Flexible decay function with tunable parameters
    
    Args:
        t: timesteps (tensor)
        tau: time constant (controls rate of descent). Lower = faster descent
        exponent: curve shape exponent. Higher = steeper curve
        curve_type: 'power' (1 - (t/tau)^exp), 'exp' (1 - e^(-t/tau)), 'log' (log(1 + t/tau))
    """
    if curve_type == 'power':
        return 1 - torch.pow(t / tau, exponent)
    elif curve_type == 'exp':
        return 1 - torch.exp(-t / tau)
    elif curve_type == 'log':
        return torch.log(1 + t / tau) / torch.log(1 + torch.tensor(101) / tau)

timesteps = torch.arange(1, 101)

# Try different time constants and exponents
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

configs = [
    {'tau': 10, 'exponent': 5.0, 'label': 'tau=10, exp=5.0'},
    {'tau': 20, 'exponent': 5.0, 'label': 'tau=20, exp=5.0'},
    {'tau': 10, 'exponent': 2.0, 'label': 'tau=10, exp=2.0'},
    {'tau': 0.5, 'exponent': 10, 'label': 'tau=0.5, exp=10'},
]

for idx, config in enumerate(configs):
    decay = decay_function(timesteps, tau=config['tau'], exponent=config['exponent'])
    axes[idx].plot(timesteps, decay, linewidth=2)
    axes[idx].set_xlabel('Timesteps')
    axes[idx].set_ylabel('Decay Value')
    axes[idx].set_title(config['label'])
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
